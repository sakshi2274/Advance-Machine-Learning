```python
# Calculate Reward
# Define actions
actions = ["Left", "Right"]

# Give reward for each action
reward = {
    "Left": -1,
    "Right": 10
}

# Print rewards
for action in actions:
    print("Action:", action)
    print("Reward:", reward[action])
```

```python
# Calculate Discounted Reward
# Rewards received
rewards = [10, 5, 2]

# Discount factor
gamma = 0.9

discounted_reward = 0

for i in range(len(rewards)):
    discounted_reward += (gamma ** i) * rewards[i]

print("Rewards:", rewards)
print("Discounted Reward:", discounted_reward)
```

```python
# Calculate Optimal Q Values
import numpy as np

# Q Table
Q = np.array([
    [0, 5],
    [2, 8],
    [4, 7]
])

print("Q Table")
print(Q)

print("\nOptimal Q Value for each State")

for i in range(len(Q)):
    print("State", i, "=", np.max(Q[i]))
```

```python
#Selecting Optimal Action
import numpy as np

Q = np.array([
    [10, 25],
    [40, 30],
    [18, 45]
])

print("Q Table")

print(Q)

print("\nBest Action")

for state in range(len(Q)):

    best_action = np.argmax(Q[state])

    print("State", state,
          "Best Action =", best_action)
```
