```python
from itertools import combinations
```

```python
#Input dataset (transactions)
transactions = [
    ['Milk', 'Bread'],
    ['Milk', 'Paper'],
    ['Bread', 'Paper'],
    ['Milk', 'Bread', 'Paper']
]
```

```python
#Set minimum support
min_support = 0.5
```

```python
# Get unique items
def get_items(transactions):
    items = []
    for t in transactions:
        for item in t:
            if item not in items:
                items.append(item)
    return items
```

```python
#Calculate support
def calculate_support(itemset):
    count = 0
    for t in transactions:
        if set(itemset).issubset(set(t)):
            count += 1
    return count / len(transactions)
```

```python
# Generate 1-itemsets
items = get_items(transactions)
current_itemsets = [[item] for item in items]
```

```python

frequent_itemsets = []
level = 1

while current_itemsets:
    print(f"\nLevel {level} Itemsets:", current_itemsets)

    next_level = []

    for itemset in current_itemsets:
        support = calculate_support(itemset)
        print(f"{itemset} -> Support: {support}")

        if support >= min_support:
            frequent_itemsets.append((itemset, support))
            next_level.append(itemset)

    # Generate next level combinations
    new_itemsets = []
    for i in range(len(next_level)):
        for j in range(i+1, len(next_level)):
            union = sorted(list(set(next_level[i]) | set(next_level[j])))

            # IMPORTANT FIX
            if len(union) == level + 1 and union not in new_itemsets:
                new_itemsets.append(union)

    # 🔴 MAIN STOP CONDITION
    if not new_itemsets:
        break

    current_itemsets = new_itemsets
    level += 1
```

```python
# Output
print("\nFrequent Itemsets:")
for itemset, sup in frequent_itemsets:
    print(itemset, "->", sup)
```
